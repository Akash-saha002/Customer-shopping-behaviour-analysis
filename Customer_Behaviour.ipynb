```python
import pandas as pd
```

```python
df = pd.read_csv("customer_shopping_behavior.csv")
df.head()
```

```python
df.isnull().sum()
```

```python
df["Review Rating"] = df.groupby("Category")["Review Rating"].transform(lambda x: x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
# create a column age group 
labels  = ["young", "adult", "middle-age", "senior"]
df["age_group"] = pd.qcut(df["Age"], q= 4, labels= labels)
```

```python
df[["Age", "age_group"]].head(10)
```

```python
# create a column of purchase frequency

frequency_mapping = {
    "Fortnightly" : 14, 
    "Weekly" : 7,
    "Monthly" : 30,
    "Quatarly" : 90,
    "Bi_Weekly" : 14,
    "Annually" : 365,
    "Every 3 Months" : 90
}
df["purchase_frequency"] = df["Frequency of Purchases"].map(frequency_mapping)
```

```python
df[["purchase_frequency", "Frequency of Purchases"]].head(6)
```

```python
df.to_csv("customer_shopping_behaviour_cleaned.csv", index=False)
```
